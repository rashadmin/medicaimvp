```python
import time
import json
import os
from langchain_google_genai import GoogleGenerativeAIEmbeddings

documents = df['text'].to_list()

CHECKPOINT_FILE = "embeddings_checkpoint.json"

def load_checkpoint():
    """Load existing embeddings from checkpoint file."""
    if os.path.exists(CHECKPOINT_FILE):
        with open(CHECKPOINT_FILE, "r") as f:
            data = json.load(f)
        print(f"🔄 Resuming from checkpoint: {len(data['embeddings'])} docs already embedded")
        return data["embeddings"], data["last_index"]
    return [], 0

def save_checkpoint(all_embeddings, last_index):
    """Save embeddings to checkpoint file."""
    with open(CHECKPOINT_FILE, "w") as f:
        json.dump({"embeddings": all_embeddings, "last_index": last_index}, f)

def embed_documents(documents: list[str], batch_size: int = 30) -> list[list[float]]:
    """Embed documents with rate limit handling + checkpoint saving."""
    embeddings_model = GoogleGenerativeAIEmbeddings(
        model="models/gemini-embedding-001",
        task_type="retrieval_document",
        google_api_key="AIzaSyAmvTxhOhadlUZM4ME0ZIEPevpP7UL-9iM"
    )

    # load from checkpoint if exists
    all_embeddings, start_index = load_checkpoint()
    total_batches = (len(documents) + batch_size - 1) // batch_size

    for i in range(start_index, len(documents), batch_size):
        batch = documents[i : i + batch_size]

        try:
            batch_embeddings = embeddings_model.embed_documents(batch)
            all_embeddings.extend(batch_embeddings)

            # ✅ save immediately after each batch
            save_checkpoint(all_embeddings, i + batch_size)
            print(f"✅ Embedded & saved {min(i + batch_size, len(documents))} / {len(documents)} docs")

        except Exception as e:
            print(f"❌ Error at batch {i // batch_size + 1}: {e}")
            print("⏳ Waiting 60s before retrying...")
            time.sleep(60)
            # retry once
            batch_embeddings = embeddings_model.embed_documents(batch)
            all_embeddings.extend(batch_embeddings)
            save_checkpoint(all_embeddings, i + batch_size)
            print(f"✅ Retry successful, saved {min(i + batch_size, len(documents))} / {len(documents)} docs")

        if i + batch_size < len(documents):
            print(f"⏳ Rate limit pause... (batch {i // batch_size + 1}/{total_batches})")
            time.sleep(62)

    # ✅ delete checkpoint once fully done
    if os.path.exists(CHECKPOINT_FILE):
        os.remove(CHECKPOINT_FILE)
        print("🗑️ Checkpoint file deleted (all done!)")

    return all_embeddings


vectors = embed_documents(documents)
print(f"Total embeddings: {len(vectors)}")
print(f"Embedding dimension: {len(vectors[0])}")
```

```python
def open_read_pdf(filename):
    docs = fitz.open(filename)
    pages = [{'source':docs.name.split('/')[-1],'page_number':page+1,'char_count':len(content.get_text()),'word_count':len(content.get_text().split(' ')),
            'token_count':len(content.get_text())/4,'sentence_count':len(content.get_text().split('.')),'text':content.get_text().replace('\n',' ').strip(),}
             for page,content in enumerate(docs)]
    return pages
def create_sources():# add filename varible instead of hardcoding it
    files = glob('/content/drive/MyDrive/tax/*')
    all_sources = []
    for file in files:
        all_sources.extend(open_read_pdf(file))
    return all_sources

def further_editing(all_sources):
    nlp = English()
    config = {"punct_chars":[';','.','?']}
    nlp.add_pipe("sentencizer",config=config)
    for item in all_sources:
        sentences = list(nlp(item['text']).sents)
        sentences = [str(sentence) for sentence in sentences]
        item['sentences'] = sentences
        item['spacy_sentences_count']= len(sentences)
        item['sentence_chunks'] = chunking(sentences)
        item['chunks_count'] = len(item['sentence_chunks'])
    return all_sources

def chunking(sentence,slice_size=5):
    return [sentence[i:i+slice_size] for i in range(0,len(sentence),slice_size)]

def page_to_chunk(all_sources):
    sentences = []
    for source in all_sources:
        for sentence in source['sentence_chunks']:
            key_map = {'source':source['source'],'page_number':source['page_number']}
            key_map['sentence_count'] = len(sentence)
            joined_sentence_chunk = "".join(sentence).replace("  ", " ").strip()
            joined_sentence_chunk = re.sub(r'\.([A-Z])', r'. \1', joined_sentence_chunk)
            key_map['sentence'] = joined_sentence_chunk
            key_map['token_count'] = len(key_map['sentence'])/4
            key_map['char_count'] = len(key_map['sentence'])
            key_map['word_count'] = len(key_map['sentence'].split(' '))
            sentences.append(key_map)
    return sentences
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
import faiss, numpy as np, pickle

# 1. Load FAISS index + df
index = faiss.read_index("index.faiss")
df = pd.read_pickle("documents.pkl")

# 2. Embed query
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001",
    task_type="retrieval_query", 
            google_api_key="AIzaSyAmvTxhOhadlUZM4ME0ZIEPevpP7UL-9iM"
    # query not document!
)
query = "CPR steps for adult cardiac arrest"
query_vector = np.array(embeddings.embed_query(query)).astype('float32').reshape(1, -1)

# 3. Search FAISS
top_k = 5
distances, indices = index.search(query_vector, top_k)

# 4. Link back to original df
results = df.iloc[indices[0]].copy()
results['distance'] = distances[0]
print(results[['text', 'distance']])  # replace 'text' with your column name
```
